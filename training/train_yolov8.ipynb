### YOLOv8 training

```python
from ultralytics import YOLO

# Load a YOLOv8 model
model = YOLO('yolov8s.pt')  # 's' = small, for a good balance between speed and accuracy

# Train the model
model.train(
    data='/content/drive/MyDrive/pig_dataset/data.yaml',   # <-- Replace with actual path
    epochs=50,
    imgsz=800,
    batch=8,
    name='pig_behavior',
    device=0,
    workers=2,
    save=True,           # Ensures model checkpoints are saved
    save_period=10       # Optional: saves every 10 epochs
)
```

### Further training the best.pt checkpoint for more Accuracy

```python
from ultralytics import YOLO

# Load the previously trained model's checkpoint
model = YOLO('/content/best_v3.pt')

# Continue training
model.train(
    data='/content/drive/MyDrive/pig_dataset/data.yaml',  # Replace with actual path
    epochs=30,                         # Additional epochs
    imgsz=800,                         # Match previous image size
    batch=8,                           # Adjust for your GPU
    name='yolov8_pig_behavior_v3',     # New experiment name
    device=0,
    workers=2,
    save=True
)
```

# Inference

```python
from ultralytics import YOLO
import cv2

# Load the trained model
#model = YOLO('/content/runs/detect/yolov8_pig_behavior_v2/weights/best.pt')  # Replace with path to your model
model = YOLO('/content/best.pt')
# Choose input file
source = '/content/pic (1).jpg'  # Replace with image.jpg or video.mp4

# Run prediction
results = model.predict(
    source=source,
    imgsz=800,        # Match your training size
    conf=0.5,         # Confidence threshold
    save=True,        # Saves output to 'runs/predict/'
    show=True         # Shows output if running locally
)
```
